```python
# Install tiktoken library [GPT models only]: pip install tiktoken
import tiktoken

# encoding = tiktoken.encoding_for_model("gpt-4.1-mini")
encoding = tiktoken.encoding_for_model("gpt-5")
# encoding = tiktoken.encoding_for_model("gpt-5-mini")

tokens = encoding.encode("Hi my name is Rahul and I love my INDIA. I like Chai Samosa.")
print(len(tokens))
print(tokens)
```

```python
# Install tiktoken library: pip install transformers
from transformers import AutoTokenizer

# tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen3-8B")
tokenizer = AutoTokenizer.from_pretrained("mistralai/Mistral-7B-v0.3")
# tokenizer = AutoTokenizer.from_pretrained("deepseek-ai/DeepSeek-V3")

tokens = tokenizer.encode("Hi my name is Rahul and I love my INDIA.")

print(tokens)
print(len(tokens))
```
