```python
 #Key must be lowercase in JSON
ticket_prices = {"goa": "$99", "paris": "$899", "tokyo": "$1400", "dubai": "$199"}
```

```python
price = ticket_prices.get("Paris".lower(), "Unknown ticket price")
print(price)
```

```python
def get_ticket_price(destination_city):
    print(f"Tool called for city {destination_city}")
    price = ticket_prices.get(destination_city.lower(), "Unknown ticket price")
    return f"The price of a ticket to {destination_city} is {price}"
get_ticket_price("London")
```

```python
students = [
    {
        "name": "Ram",
        "age": 20
    },
    {
        "name": "Amit",
        "age": 25
    },
    {
        "name": "Neha",
        "age": 22
    }
]

print(json.dumps(students, indent=4))

# Loop through

for s in students:
    print(s["name"], s["age"])
    print(s.get("name"),s["age"])
```

```python
json_string = '''
{
    "company":"INCAPP",
    "courses":[
        {
            "name":"Python",
            "duration":"2 Months"
        },
        {
            "name":"Java",
            "duration":"3 Months"
        }
    ]
}
'''

data = json.loads(json_string)

print(data["company"])

for course in data["courses"]:
    print(course["name"], course["duration"])
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

users = response.json()

print(type(users))

# Display users

for user in users:
    print(type(user))
    print(user["name"], "-", user["email"])
```
