```python
from reportlab.platypus import SimpleDocTemplate, Paragraph, HRFlowable, Spacer
from reportlab.lib.styles import getSampleStyleSheet

pdf = SimpleDocTemplate("First.pdf")
#pdf =SimpleDocTemplate(D:/spring/First.pdf)

styles = getSampleStyleSheet()

story = []

story.append(
    Paragraph("INCAPP Coding Institute", styles["Title"])
)
story.append(HRFlowable(width="100%"))
story.append(
    Paragraph("AI Engineering", styles["Heading1"])
)
story.append(Paragraph("<br/>", styles["BodyText"]))
story.append(Spacer(1, 20))
story.append(
    Paragraph(
        "Learn <i>Java</i>, <u>Python</u>, <b>AI</b> and Full Stack Development.",styles["BodyText"]
    )
)

pdf.build(story)

print("PDF Created Successfully")
```
