```python
import gradio as gr

# Function to calculate Simple Interest
def simple_interest(principal, rate, time):
    si = (principal * rate * time) / 100
    total_amount = principal + si
    return f"Simple Interest: ₹{si:.2f}\nTotal Amount: ₹{total_amount:.2f}"

# Gradio Interface
demo = gr.Interface(
    fn=simple_interest,
    inputs=[
        gr.Number(label="Principal Amount"),
        gr.Number(label="Rate of Interest"),
        gr.Number(label="Time (Years)")
    ],
    outputs=gr.Textbox(label="Result"),
    title="Simple Interest Calculator"

)

demo.launch()
```
